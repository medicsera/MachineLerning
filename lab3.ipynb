```python
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("la-crimes-sample.csv")

"""Сколько строк и столбцов в таблице?"""
rows, cols = df.shape
print(f"Строчек: {rows}, столбцов: {cols}\n")

""" Каковы названия столбцов? Какие типы данных у столбцов?"""
print(f"Инфа: {df.info()}\n")

""" Сколько в каждом из них уникальных значений?"""
print(f"Уникальный значений: {df.nunique()}\n")

""" Сколько пропущенных значений?"""
print(f"Пропущенных значений: {df.isnull().sum()}")

""" Верно ли, что женщины чаще оказываются жертвами по сравнению с мужчинами?"""
sex_value = df["Victim Sex"].value_counts()
print(f"Количество жертв Ж: {sex_value.get("F",0)} против жертв мужчин: {sex_value.get("M",0)}")

""" Определите 10 самых распространенных преступлений в Лос-Анджелесе.
Постройте график."""
top_10_crimes = df["Crime Code Description"].value_counts().head(10)
print(f"10 распространённых  преступлений: {top_10_crimes}")

plt.figure(figsize=(12,8))

top_10_crimes.sort_values().plot(kind="barh")

plt.title("Топ 10 преступлений")
plt.xlabel("Количество случаев")
plt.ylabel("Тип преступления")

plt.show()

""" От каких преступлений чаще страдают женщины, а от каких мужчины?"""
crimes_with_sex_list =  df.groupby(["Crime Code Description","Victim Sex"]).size().unstack(fill_value=0)

men_list = crimes_with_sex_list[crimes_with_sex_list['M'] > crimes_with_sex_list["F"]].index.tolist()
print(f"Страдают мужчины: {men_list}")

women_list = crimes_with_sex_list[crimes_with_sex_list['F'] > crimes_with_sex_list["M"]].index.tolist()
print(f"Страдают женщины: {women_list}")

""" Люди какого происхождения чаще всего подвергаются преступлениям?"""
descent = df["Victim Descent"].value_counts()
print(f"Чаще всего подвергались по происхождению: {descent}")

""" Отсортируйте районы, по количеству преступлений. Постройте график,
показывающий самые безопасный и опасный районы."""
area_count = df['Area Name'].value_counts()
print(f"От опасных к безопасным: {area_count}")

plt.figure(figsize=(12,8))

area_count.sort_values().plot(kind="barh")

plt.title("Кол-во преступление в районах")
plt.xlabel("Количество преступлений")
plt.ylabel("Район")

plt.show()
```

```python
import pandas as pd

"""1. Загрузите датафрейм из файла polit.csv и сохраните его в переменную polit. Если в
датафрейме есть строки с пропущенными значениями, то удалите их."""
polit = pd.read_csv('polit.csv', sep=',')

cols_to_fix = ['fh09', 'polity09', 'gini', 'fparl08', 'corr0509']

for col in cols_to_fix:
    polit[col] = (
        polit[col]
        .astype(str)
        .str.replace(',', '.', regex=False)
        .replace('nan', pd.NA)
        .replace('', pd.NA)
    )
    polit[col] = pd.to_numeric(polit[col], errors='coerce')

polit_clean = polit.dropna(subset=['fh09', 'polity09', 'gini', 'fparl08', 'corr0509']).copy()

polit = polit_clean

print(polit)
```

6. Добавьте в датафрейм polit столбец fh_status, в котором будут храниться типы стран
в зависимости от значения индекса Freedom House (значения free, partly free, not free).
Соответствие значений fh09 с типами стран см. здесь
https://freedomhouse.org/sites/default/files/2020-02/Methodology_FIW_2016.pdf (Table
3 на последней странице).

```python
def classify_fh_status(fh):
    if 1.0 <= fh <= 2.5:
        return 'free'
    elif 3.0 <= fh <= 5.0:
        return 'partly free'
    elif 5.5 <= fh <= 7.0:
        return 'not free'
    else:
        return None

polit['fh_status'] = polit['fh09'].apply(classify_fh_status)
print(polit[['fh09','fh_status']])
```

1. Вывести все доступные платформы, на которых выпускалась хотя-бы одна игра;

```python
vgsales = pd.read_csv('vgsales.csv')
metacritic = pd.read_csv('metacritic_games.csv')

vgsales.columns = vgsales.columns.str.lower()
metacritic.columns = metacritic.columns.str.lower()

platforms = sorted(vgsales['platform'].dropna().unique())
print("1. Доступные платформы:")
for p in platforms:
    print(p)
```

2. Добавить к копии набора данных столбец metacritic_rating и заполнить его данными
из набора данных Metacritic all time games stats
(https://www.kaggle.com/skateddu/metacritic-all-time-games-stats);


```python
vgsales_with_meta = vgsales.copy()

vgsales_with_meta['name_clean'] = vgsales_with_meta['name'].str.lower().str.strip()
metacritic['name_clean'] = metacritic['name'].str.lower().str.strip()

merged = pd.merge(
    vgsales_with_meta,
    metacritic[['name_clean', 'metascore']],
    on='name_clean',
    how='left'
)

merged = merged.drop(columns=['name_clean'])
merged = merged.rename(columns={'metascore': 'metacritic_rating'})

print(merged)
```

5. Вывести жанры игр с количеством игр, которые принадлежат соответствующему
жанру в виде "<жанр> - <количество игр>" для всех жанров, которые содержат в
своем названии не менее 3 различных гласных.

```python
vowels = set('aeiou')

def has_at_least_3_vowels(genre):
    if not isinstance(genre, str):
        return False
    return len(set(genre.lower()) & vowels) >= 3

genre_counts = merged['genre'].value_counts()

result = {}
for genre, count in genre_counts.items():
    if has_at_least_3_vowels(genre):
        result[genre] = count

for genre in sorted(result):
    print(f"{genre} - {result[genre]}")
```
