```python
import pandas as pd
import numpy as np
from statsmodels.stats.weightstats import _tconfint_generic as t_confidence_interval

df = pd.read_csv('water.txt', sep='\t')

df.columns = ['location', 'town', 'mortality', 'hardness']
df['location'] = df['location'].replace({'Sout': 'South', 'Nort': 'North'})

print("=== Описательные статистики ===")
print(df.groupby('location')[['mortality', 'hardness']].describe())
print("\n")

def get_95_ci(series):
    mean = series.mean()
    std = series.std()
    n = len(series)
    
    mean_std = std / np.sqrt(n)
    dof = n - 1
    
    lower, upper = t_confidence_interval(mean, mean_std, dof, 0.05, 'two-sided')
    return lower, upper

print("=== 95% Доверительные интервалы для смертности ===")
for region in ['South', 'North']:
    subset = df[df['location'] == region]['mortality']
    ci = get_95_ci(subset)
    print(f"{region}: ({ci[0]:.4f}, {ci[1]:.4f})")

print("\n=== 95% Доверительные интервалы для жесткости воды ===")
for region in ['South', 'North']:
    subset = df[df['location'] == region]['hardness']
    ci = get_95_ci(subset)
    print(f"{region}: ({ci[0]:.4f}, {ci[1]:.4f})")
```
